```python
import pandas as pd
from scipy.stats.mstats import gmean, hmean
import matplotlib.pyplot as plt

# Data dimasukkan 
my_dataset = pd.DataFrame({
    'Zr': [300, 320, 350, 370, 400, 420, 380, 360, 340, 390]
})

# Menghitung rata-rata
a_mean = my_dataset.Zr.mean()
g_mean = gmean(my_dataset['Zr'])
h_mean = hmean(my_dataset['Zr'])

print('-------')
print('arithmetic mean')
print("{0:.1f} [ppm]".format(a_mean))
print('-------')

print('geometric mean')
print("{0:.1f} [ppm]".format(g_mean))
print('-------')

print('harmonic mean')
print("{0:.1f} [ppm]".format(h_mean))
print('-------')

# Membuat grafik
fig, ax = plt.subplots()
ax.hist(my_dataset.Zr, bins='auto', density=True, edgecolor='k', alpha=0.8)

ax.axvline(a_mean, color='purple', label='Arithmetic mean', linewidth=2)
ax.axvline(g_mean, color='orange', label='Geometric mean', linewidth=2)
ax.axvline(h_mean, color='green', label='Harmonic mean', linewidth=2)

ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')
ax.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membuat data
data = {
    'Zr': [100, 120, 150, 180, 200, 220, 240, 260, 280, 280, 280,
           300, 320, 340, 360, 380, 400, 420]
}

# Mengubah menjadi DataFrame
my_dataset = pd.DataFrame(data)

# Menghitung histogram secara numerik (tanpa plot)
hist, bin_edges = np.histogram(my_dataset['Zr'], bins=10, density=True)

# Menentukan nilai modus (nilai yang paling sering muncul berdasarkan histogram)
modal_value = (bin_edges[hist.argmax()] + bin_edges[hist.argmax()+1]) / 2

# Menampilkan nilai modus
print('modal value: {0:.0f} [ppm]'.format(modal_value))

# Membuat grafik histogram
fig, ax = plt.subplots()

ax.hist(my_dataset.Zr, bins=10, density=True, edgecolor='k',
        label="Measurements Hist", alpha=0.8)

# Menambahkan garis vertikal untuk menunjukkan nilai modus
ax.axvline(modal_value, color="orange", label="Modal value", linewidth=2)

# Memberi label pada sumbu
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')

# Menampilkan legenda
ax.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Membuat data 
data = {
    'Zr': [100, 130, 160, 200, 230, 260, 290, 320, 350, 380, 410, 440]
}

# Mengubah menjadi DataFrame
my_dataset = pd.DataFrame(data)

# Menghitung variance (ragam) dan standard deviation (simpangan baku)
variance = my_dataset['Zr'].var()
stddev = my_dataset['Zr'].std()

print('-------')
print('Variance')
print("{0:.0f} [square ppm]".format(variance))
print('-------')
print('Standard Deviation')
print("{0:.0f} [ppm]".format(stddev))
print('-------')

# Membuat histogram
fig, ax = plt.subplots()

ax.hist(my_dataset.Zr, bins=10, density=True, edgecolor='k',
        label='Measurements Hist')

# Menambahkan garis mean - 1 standar deviasi
ax.axvline(my_dataset['Zr'].mean() - stddev,
           color='purple', label=r'mean - 1$\sigma$', linewidth=2)

# Menambahkan garis mean + 1 standar deviasi
ax.axvline(my_dataset['Zr'].mean() + stddev,
           color='green', label=r'mean + 1$\sigma$', linewidth=2)

# Menandai area mean ± 1 standar deviasi
ax.axvspan(my_dataset['Zr'].mean() - stddev,
           my_dataset['Zr'].mean() + stddev,
           alpha=0.1, color='orange',
           label=r'mean $\pm$ 1$\sigma$')

# Label sumbu
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')

# Menampilkan legenda
ax.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membuat data 
data = {
    'Zr': [95, 120, 140, 160, 180, 200, 220, 240, 260, 280,
           300, 320, 340, 360, 380, 400]
}

# Mengubah menjadi DataFrame
my_dataset = pd.DataFrame(data)

# Menghitung kuartil pertama (Q1 / 25%)
q1 = np.percentile(my_dataset.Zr, 25, method='midpoint')

# Menghitung kuartil ketiga (Q3 / 75%)
q3 = np.percentile(my_dataset.Zr, 75, method='midpoint')

# Menghitung IQR (selisih Q3 dan Q1)
iqr = q3 - q1  # Interquartile range (IQR)

print('-------')
print('Interquartile range (IQR): {0:.0f} [ppm]'.format(iqr))
print('-------')

# Membuat histogram
fig, ax = plt.subplots()

ax.hist(my_dataset.Zr, bins='auto', density=True, edgecolor='k',
        label='Measurements Hist')

# Garis untuk Q1
ax.axvline(q1, color='purple', label='Q1', linewidth=2)

# Garis untuk Q3
ax.axvline(q3, color='green', label='Q3', linewidth=2)

# Area IQR (antara Q1 dan Q3)
ax.axvspan(q1, q3, alpha=0.1, color='orange',
           label='Interquartile range (IQR)')

# Label sumbu
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')

# Menampilkan legenda
ax.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy.stats import skew

# Membuat data 
data = {
    'Zr': [100, 120, 140, 160, 180, 200, 220, 240, 260, 280,
           300, 350, 400, 450, 500]
}

# Mengubah menjadi DataFrame
my_dataset = pd.DataFrame(data)

# Menghitung mean (rata-rata)
a_mean = my_dataset.Zr.mean()

# Menghitung median (nilai tengah)
median = my_dataset.Zr.median()

# Menghitung histogram untuk mencari modus
hist, bin_edges = np.histogram(my_dataset['Zr'], bins=10, density=True)

# Menentukan modus (nilai yang paling sering muncul)
modal_value = (bin_edges[hist.argmax()] + bin_edges[hist.argmax()+1]) / 2

# Menghitung standar deviasi
standard_deviation = my_dataset['Zr'].std()

# Pearson’s first coefficient of skewness (berdasarkan modus)
a1 = (a_mean - modal_value) / standard_deviation 

# Pearson’s second coefficient (berdasarkan median)
a2 = 3 * (a_mean - median) / standard_deviation

# Fisher-Pearson skewness (menggunakan fungsi scipy)
g1 = skew(my_dataset['Zr'])

print('-------')
print("Pearson's first coefficient of skewness: {:.2f}".format(a1))
print("Pearson's 2nd moment of skewness: {:.2f}".format(a2))
print("Fisher-Pearson's coefficient of skewness: {:.2f}".format(g1))
print('-------')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Membuat data  
data = {
    'Zr': [100, 120, 140, 160, 180, 200, 220, 240, 260, 280,
           300, 320, 340, 360, 380, 400, 420]
}

# Mengubah menjadi DataFrame
my_dataset = pd.DataFrame(data)

# Membuat area plot
fig, ax = plt.subplots()

# Mengatur tampilan outlier (titik di luar data normal)
my_flierprops = dict(markerfacecolor="#288974",
                     markeredgecolor="#d049c3",
                     marker='o')

# Mengatur tampilan garis median
my_medianprops = dict(color="#b4c0a7", linewidth=2)

# Mengatur tampilan kotak boxplot
my_boxprops = dict(facecolor="#4d5764",
                   edgecolor="#a37fa0")

# Membuat boxplot dari data Zr
ax.boxplot(my_dataset.Zr,
           patch_artist=True,
           notch=True,
           flierprops=my_flierprops,
           medianprops=my_medianprops,
           boxprops=my_boxprops)

# Label sumbu Y
ax.set_ylabel('Zr [ppm]')

# Mengatur label sumbu X
ax.set_xticks([1])
ax.set_xticklabels(['all Epochs'])

plt.show()
```
